# InstaSHAP Analysis Notebook

This notebook mirrors the reproducibility pipeline implemented in the project package. It demonstrates dataset loading via `ucimlrepo`, experiment execution, and access to saved plots/tables.

```python
from pathlib import Path
import yaml

from instashap_project.data.loaders import load_bike_sharing, load_covertype, load_adult_income
from instashap_project.experiments.bike_sharing import run as run_bike
from instashap_project.experiments.covertype import run as run_covertype
from instashap_project.experiments.adult_income import run as run_adult
from instashap_project.reports.generate_report import generate_full_report

project_root = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
config = yaml.safe_load((project_root / 'config.yaml').read_text(encoding='utf-8'))
config['global']['fast_dev_run'] = True
config
```

```python
bike = load_bike_sharing()
covertype = load_covertype(max_rows=1000, seed=42)
adult = load_adult_income(max_rows=1000, seed=42)

print('Bike:', bike.features.shape, bike.target.shape)
print('Covertype:', covertype.features.shape, covertype.target.shape)
print('Adult:', adult.features.shape, adult.target.shape)
bike.features.head()
```

```python
bike_result = run_bike(config=config, selected_model='instashap')
adult_result = run_adult(config=config, selected_model='gam')
print(bike_result)
print(adult_result)
```

```python
report_path = generate_full_report(project_root)
report_path
```
